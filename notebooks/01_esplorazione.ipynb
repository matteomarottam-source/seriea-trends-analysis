```python
import matplotlib.pyplot as plt
```

```python
df['FTR'].value_counts().plot(kind='bar', color=['steelblue', 'gray', 'indianred'])
plt.title('Risultati Serie A 2025/2026')
plt.savefig('../images/risultati_serie_a.png', bbox_inches='tight')
plt.show()
```

```python
df.groupby('HomeTeam')['FTHG'].mean().sort_values(ascending=False)
```

```python
df.groupby('AwayTeam')['FTAG'].mean().sort_values(ascending=False).plot(kind='bar', color='indianred')
plt.title('Media Goal in Trasferta per Squadra')
plt.xlabel('Squadra')
plt.ylabel('Media Goal')
plt.xticks(rotation=45, ha='right')
plt.savefig('../images/media_goal_trasferta.png', bbox_inches='tight')
plt.show()
```
